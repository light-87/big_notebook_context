```python
# Directory Scanner for Jupyter Notebook

import os
from collections import defaultdict
from IPython.display import display, Markdown


def scan_directory(path: str):
    """
    Recursively scans `path`, builds a tree-like ASCII structure,
    and groups files by extension.

    Returns:
      - structure_lines: List of strings (each line of the tree)
      - file_types: Dict mapping extension to list of file paths
    """
    file_types = defaultdict(list)
    structure_lines = []

    def recurse(current_path: str, prefix: str = ""):
        try:
            entries = sorted(os.listdir(current_path))
        except PermissionError:
            structure_lines.append(f"{prefix}└── [Permission Denied]")
            return

        for idx, entry in enumerate(entries):
            full_path = os.path.join(current_path, entry)
            connector = "└── " if idx == len(entries) - 1 else "├── "
            structure_lines.append(f"{prefix}{connector}{entry}")

            if os.path.isdir(full_path):
                spacer = "    " if idx == len(entries) - 1 else "│   "
                recurse(full_path, prefix + spacer)
            else:
                ext = os.path.splitext(entry)[1].lower() or 'no_extension'
                file_types[ext].append(full_path)

    structure_lines.append(f"**Scanning:** `{path}`")
    recurse(path)
    return structure_lines, file_types
```

```python
# Example usage in a notebook:

# 1. Set the path you want to scan
directory_path = "./"

# 2. Call the function
tree, grouped = scan_directory(directory_path)

# 3. Display the tree
display(Markdown("\n".join(tree)))

# 4. Display files by type
for ext, files in grouped.items():
    display(Markdown(f"### {ext} — {len(files)} files"))
    for f in files:
        display(Markdown(f"- `{f}`"))

# Uncomment and modify `directory_path` above to run.
```

```python
# Enhanced Directory Scanner for Jupyter Notebook

import os
import re
from collections import defaultdict
from IPython.display import display, Markdown, FileLink


def scan_directory(path: str):
    """
    Recursively scans `path`, builds a tree-like ASCII structure,
    and groups files by extension.

    Returns:
      - structure_lines: List of strings (each line of the tree)
      - file_types: Dict mapping extension to list of file paths
    """
    file_types = defaultdict(list)
    structure_lines = []

    def recurse(current_path: str, prefix: str = ""):
        try:
            entries = sorted(os.listdir(current_path))
        except PermissionError:
            structure_lines.append(f"{prefix}└── [Permission Denied]")
            return

        for idx, entry in enumerate(entries):
            full_path = os.path.join(current_path, entry)
            connector = "└── " if idx == len(entries) - 1 else "├── "
            structure_lines.append(f"{prefix}{connector}{entry}")

            if os.path.isdir(full_path):
                spacer = "    " if idx == len(entries) - 1 else "│   "
                recurse(full_path, prefix + spacer)
            else:
                ext = os.path.splitext(entry)[1].lower() or 'no_extension'
                file_types[ext].append(full_path)

    structure_lines.append(f"**Scanning:** `{path}`")
    recurse(path)
    return structure_lines, file_types


def write_markdown_reports(path: str,
                           tree_file: str = "directory_tree.md",
                           types_file: str = "files_by_type.md") -> tuple:
    """
    Generates two Markdown files:
      - `tree_file`: Nested Markdown list of directory structure
      - `types_file`: Files grouped by extension

    Returns:
      (tree_file, types_file)
    """
    tree, grouped = scan_directory(path)

    # Write directory tree as nested Markdown list
    with open(tree_file, 'w', encoding='utf-8') as tf:
        tf.write(f"# Directory Tree of `{path}`\n\n")
        for line in tree:
            if line.startswith("**Scanning**"):
                continue
            match = re.match(r"(?P<pref>[ │]*)(?:├── |└── )(?P<name>.*)", line)
            if match:
                level = len(match.group('pref')) // 4
                tf.write('  ' * level + f"- {match.group('name')}\n")
            else:
                tf.write(f"- `{line}`\n")

    # Write files grouped by type
    with open(types_file, 'w', encoding='utf-8') as ff:
        ff.write(f"# Files by Type in `{path}`\n\n")
        for ext, files in sorted(grouped.items()):
            ff.write(f"## {ext} ({len(files)})\n\n")
            for fpath in files:
                ff.write(f"- `{fpath}`\n")
            ff.write("\n")

    return tree_file, types_file

# Usage in a Jupyter notebook:

# 1. Set the directory to scan
directory_path = "./"

# 2. Generate Markdown reports
tree_md, types_md = write_markdown_reports(directory_path)

# 3. Display generated file links
display(Markdown("### Generated Reports"))
display(FileLink(tree_md))
display(FileLink(types_md))
```
