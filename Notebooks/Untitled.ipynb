```python
import json
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
from pathlib import Path

# Set up matplotlib for high-quality plots
plt.rcParams.update({
    'figure.figsize': (15, 10),
    'font.size': 14,
    'axes.titlesize': 16,
    'axes.labelsize': 14,
    'xtick.labelsize': 12,
    'ytick.labelsize': 12,
    'legend.fontsize': 12,
    'figure.titlesize': 18,
    'lines.linewidth': 2,
    'lines.markersize': 6
})

class TrainingPlotter:
    def __init__(self):
        self.data = None
    
    def load_data(self, file_path=None):
        """Load training data from JSON file"""
        if file_path is None:
            # Look for training_history.json in current directory
            possible_files = [
                'training_history.json',
                'history.json',
                'training_log.json'
            ]
            
            file_path = None
            for f in possible_files:
                if os.path.exists(f):
                    file_path = f
                    break
            
            if file_path is None:
                print("❌ No training_history.json found in current directory.")
                print("📁 Please specify the file path or place your file in the current directory.")
                return False
        
        try:
            with open(file_path, 'r') as f:
                self.data = json.load(f)
            print(f"✅ Successfully loaded: {file_path}")
            print(f"📊 Available metrics: {list(self.data.keys())}")
            return True
        except Exception as e:
            print(f"❌ Error loading file: {str(e)}")
            return False
    
    def create_training_plots(self, save_plots=False, dpi=300):
        """Create the training plots with improved styling"""
        if self.data is None:
            print("❌ Please load data first using load_data() method!")
            return
        
        # Common metrics mapping - adjust these based on your actual JSON keys
        metric_configs = [
            {'keys': ['loss', 'val_loss'], 'title': 'Training Loss', 'ylabel': 'Loss', 'colors': ['#d62728', '#1f77b4']},
            {'keys': ['accuracy', 'val_accuracy'], 'title': 'Training Accuracy', 'ylabel': 'Accuracy', 'colors': ['#d62728', '#1f77b4']},
            {'keys': ['f1_score', 'val_f1_score'], 'title': 'F1 Score', 'ylabel': 'F1 Score', 'colors': ['#d62728', '#1f77b4']},
            {'keys': ['precision', 'val_precision'], 'title': 'Precision', 'ylabel': 'Precision', 'colors': ['#d62728', '#1f77b4']},
            {'keys': ['recall', 'val_recall'], 'title': 'Recall', 'ylabel': 'Recall', 'colors': ['#d62728', '#1f77b4']},
            {'keys': ['auc', 'val_auc'], 'title': 'AUC', 'ylabel': 'AUC', 'colors': ['#d62728', '#1f77b4']}
        ]
        
        # Try alternative key names as well
        alternative_keys = {
            'accuracy': ['acc', 'train_accuracy', 'train_acc'],
            'val_accuracy': ['val_acc', 'validation_accuracy', 'validation_acc'],
            'loss': ['train_loss', 'training_loss'],
            'val_loss': ['validation_loss', 'valid_loss']
        }
        
        # Update metric configs with alternative keys
        for config in metric_configs:
            for i, key in enumerate(config['keys']):
                if key not in self.data and key in alternative_keys:
                    for alt_key in alternative_keys[key]:
                        if alt_key in self.data:
                            config['keys'][i] = alt_key
                            break
        
        # Filter available metrics
        available_metrics = []
        for config in metric_configs:
            if any(key in self.data for key in config['keys']):
                available_metrics.append(config)
        
        if not available_metrics:
            print("❌ No recognizable metrics found in the data!")
            print("📋 Available keys:", list(self.data.keys()))
            print("💡 You might need to adjust the metric_configs in the code")
            return
        
        # Calculate subplot layout
        n_metrics = len(available_metrics)
        n_cols = 3 if n_metrics >= 6 else 2
        n_rows = (n_metrics + n_cols - 1) // n_cols
        
        # Create figure with subplots
        fig, axes = plt.subplots(n_rows, n_cols, figsize=(6*n_cols, 5*n_rows))
        
        # Handle single subplot case
        if n_metrics == 1:
            axes = [axes]
        elif n_rows == 1:
            axes = axes.reshape(1, -1) if n_cols > 1 else [axes]
        elif n_cols == 1:
            axes = axes.reshape(-1, 1)
        
        # Plot each metric
        for idx, config in enumerate(available_metrics):
            if n_rows == 1 and n_cols == 1:
                ax = axes[0]
            elif n_rows == 1:
                ax = axes[idx]
            elif n_cols == 1:
                ax = axes[idx]
            else:
                row, col = idx // n_cols, idx % n_cols
                ax = axes[row][col]
            
            epochs = None
            
            for i, key in enumerate(config['keys']):
                if key in self.data:
                    values = self.data[key]
                    if epochs is None:
                        epochs = list(range(1, len(values) + 1))
                    
                    # Determine label and styling
                    if 'val_' in key or 'validation' in key.lower():
                        label = 'Validation'
                        linestyle = '--'
                        marker = 's'
                    else:
                        label = 'Training'
                        linestyle = '-'
                        marker = 'o'
                    
                    # Plot with improved styling
                    ax.plot(epochs, values, 
                           color=config['colors'][i], 
                           linestyle=linestyle,
                           marker=marker,
                           markersize=4,
                           label=label,
                           linewidth=2.5,
                           alpha=0.8)
            
            # Customize subplot
            ax.set_title(config['title'], fontsize=16, fontweight='bold', pad=15)
            ax.set_xlabel('Epochs', fontsize=14, fontweight='bold')
            ax.set_ylabel(config['ylabel'], fontsize=14, fontweight='bold')
            ax.grid(True, alpha=0.3, linewidth=0.8)
            ax.legend(fontsize=12, loc='best', frameon=True, fancybox=True, shadow=True)
            
            # Improve tick formatting
            ax.tick_params(axis='both', which='major', labelsize=12, width=1.2)
            
            # Set reasonable limits
            if epochs:
                ax.set_xlim(1, max(epochs))
            
            # Add border
            for spine in ax.spines.values():
                spine.set_linewidth(1.2)
        
        # Hide empty subplots
        total_subplots = n_rows * n_cols
        if n_metrics < total_subplots:
            for idx in range(n_metrics, total_subplots):
                if n_rows == 1:
                    axes[idx].set_visible(False)
                else:
                    row, col = idx // n_cols, idx % n_cols
                    axes[row][col].set_visible(False)
        
        # Adjust layout
        plt.tight_layout(pad=3.0)
        
        # Add main title
        fig.suptitle('Training Progress', fontsize=20, fontweight='bold', y=0.98)
        
        if save_plots:
            plt.savefig('training_curves_combined.png', dpi=dpi, bbox_inches='tight')
            print("💾 Saved combined plot as 'training_curves_combined.png'")
        
        # Show plot
        plt.show()
        
        # Create individual plots
        self.create_individual_plots(available_metrics, save_plots, dpi)
    
    def create_individual_plots(self, available_metrics, save_plots=False, dpi=300):
        """Create individual high-resolution plots for each metric"""
        print("\n📈 Creating individual high-resolution plots...")
        
        for config in available_metrics:
            fig, ax = plt.subplots(1, 1, figsize=(10, 6))
            
            epochs = None
            
            for i, key in enumerate(config['keys']):
                if key in self.data:
                    values = self.data[key]
                    if epochs is None:
                        epochs = list(range(1, len(values) + 1))
                    
                    # Determine styling
                    if 'val_' in key or 'validation' in key.lower():
                        label = 'Validation'
                        linestyle = '--'
                        marker = 's'
                    else:
                        label = 'Training'
                        linestyle = '-'
                        marker = 'o'
                    
                    ax.plot(epochs, values, 
                           color=config['colors'][i], 
                           linestyle=linestyle,
                           marker=marker,
                           markersize=6,
                           label=label,
                           linewidth=3,
                           alpha=0.8)
            
            # Enhanced styling for individual plots
            ax.set_title(config['title'], fontsize=18, fontweight='bold', pad=20)
            ax.set_xlabel('Epochs', fontsize=16, fontweight='bold')
            ax.set_ylabel(config['ylabel'], fontsize=16, fontweight='bold')
            ax.grid(True, alpha=0.3, linewidth=1)
            ax.legend(fontsize=14, loc='best', frameon=True, shadow=True, fancybox=True)
            
            # Enhanced tick formatting
            ax.tick_params(axis='both', which='major', labelsize=14, width=1.5)
            
            # Set limits
            if epochs:
                ax.set_xlim(1, max(epochs))
            
            # Enhanced border
            for spine in ax.spines.values():
                spine.set_linewidth(1.5)
            
            plt.tight_layout()
            
            if save_plots:
                filename = f"training_{config['title'].lower().replace(' ', '_')}.png"
                plt.savefig(filename, dpi=dpi, bbox_inches='tight')
                print(f"💾 Saved {filename}")
            
            plt.show()
        
        print("✅ All plots generated successfully!")

# Usage Instructions and Helper Functions
def quick_plot(file_path=None, save_plots=True):
    """Quick function to generate all plots"""
    plotter = TrainingPlotter()
    if plotter.load_data(file_path):
        plotter.create_training_plots(save_plots=save_plots, dpi=300)

def list_json_files():
    """List all JSON files in current directory"""
    json_files = [f for f in os.listdir('.') if f.endswith('.json')]
    if json_files:
        print("📁 JSON files found in current directory:")
        for f in json_files:
            print(f"   - {f}")
    else:
        print("❌ No JSON files found in current directory")

# Instructions
print("🚀 Training Curve Plotter Ready!")
print("\n📋 USAGE OPTIONS:")
print("1. Put your 'training_history.json' in the same folder and run:")
print("   quick_plot()")
print("\n2. Or specify the file path:")
print("   quick_plot('path/to/your/training_history.json')")
print("\n3. Or create a plotter object for more control:")
print("   plotter = TrainingPlotter()")
print("   plotter.load_data('your_file.json')")
print("   plotter.create_training_plots(save_plots=True)")
print("\n4. To see available JSON files:")
print("   list_json_files()")
print("\n💡 The plots will be optimized for dissertation quality with larger fonts")
print("💾 Set save_plots=True to save high-resolution PNG files")
```
